## 1.3. Type Conversion

Python supports two types of data type conversion:

1. Implicit Conversion
2. Explicit Conversion

### Implicit Conversion

Python automatically converts one data type to another when required.

Example:

$$
1 + 1.0
$$

Python automatically converts the integer value `1` into a float and performs the addition.

This is known as **Implicit Conversion**.

````python
num=1 + 1.0
print(num)
````


### Explicit Conversion

If we want the output as:

$$
2
$$

then we must explicitly convert the float value `1.0` into the integer value `1`.

This is known as **Explicit Conversion**.

Explicit conversion involves specifying the target data type explicitly.


````python
num=1 + int(1.0)
print(num)
````

### Dynamic Typing vs Static Typing

Languages like Python are dynamically typed, whereas C, Go, etc. are statically typed.

### Dynamic Typing

Dynamic Typing is a technique in some languages where, depending on how a value is used, the data type of the variable is dynamically and automatically assigned.

Consider the following code in Python:

`num = 65`  (Line 1)

`num = "A"`  (Line 2)

- In Line 1, variable `num` is considered to be of type `int`.
- In Line 2, its type is reassigned to `String`.










### Static Typing

Static Typing is used in some languages where the data type has to be declared before a variable is used.

Consider the following code in Go:

`var num int = 65`  (Line 1)

`num = "A"`  (Line 2)

`var name string = "A"`  (Line 3)

- Line 1 is a valid statement that declares a variable `num` of type `int`.
- Line 2 is invalid because a string value cannot be assigned to a variable already declared as type `int`.
- Line 3 is a valid statement where `name` is declared and used as a `string`.

## 2.5. Range

We can create a sequence of values in Python using `range(x,y,step)`.

It creates a sequence from `x` to `y-1` with a difference of `step` between each value.

### Parameters

- `start` - Starting number of the sequence
- `end` - Generate number up to end, but not including this number
- `step` - Difference between each number in the sequence

It is not mandatory to give `step`. The default value of `step` is `1`.

````python
start=1
end=10
step=2
for number in range (start, end, step):
    print("The current number is ", number)
for number in range(1,7,2):
    print ("The current number is ",number)

print("---------------------------")

for number in range(5,0,-1):
    print ("The current number is ",number)
````

### When to Use Loop

| `while` loop | When number of iterations are not known |

## 2.6. Break & Continue

When we want to stop a loop or break away from it, we can use the `break` statement.

When we want to skip the remaining portion of loop statements and continue with the next iteration, we can use the `continue` statement.

````python
for passenger in "A","A", "FC", "C", "FA",  "SP", "A", "A":
    if(passenger=="FC" or passenger=="FA"):
        print("No check required")
        continue
    if(passenger=="SP"):
        print("Declare emergency in the airport")
        break
    if(passenger=="A" or passenger=="C"):
        print("Proceed with normal security check")
    print("Check the person")
    print("Check for cabin baggage")
````

Given a list of integer values, write a Python program to check whether it contains same number in adjacent position. Display the count of such adjacent occurrences.

#### Sample Input / Expected Output

| Sample Input | Expected Output |
|---|---|
| `[1,1,5,100,-20,-20,6,0,0]` | `3` |
| `[10,20,30,40,30,20]` | `0` |
| `[1,2,2,3,4,4,4,10]` | `3` |

#### Test Case

| Sl. No. | Type | S/A | Test Target | Input | Expected Output | Actual Output | Result |
|---|---|---|---|---|---|---|---|
| 1 | Structural | N/A | `get_count(num_list)` | NA | NA | NA | done |
| 2 | Logical | Sample | `get_count` | `num_list-[23, 23, 6, -7, -7, 8, 0, 0]` | 3 | 3 | done |
| 3 | Actual |  | `get_count` | `num_list-[98, 97, 97, 98]` | N/A | 1 | done |


````python
def get_count(num_list):
    count=0
    list_len = len(num_list)
    # Write your logic here
    for i in range(list_len):
        if(i==list_len-1):
            break
        else:
            if(num_list[i]==num_list[i+1]):
                count=count+1
    return count

#provide different values in list and test your program
num_list=[1,1,5,100,-20,-20,6,0,0]
print(get_count(num_list))
````

Like a list, a tuple can also store a sequence of elements, but the value of the elements cannot be changed.

## 3.3. Dictionary

### 3.3.1. Introduction

A dictionaries can be used to store an unordered collection of key-value pairs.

The key should be unique and can be of any immutable data type.

Like lists, dictionaries are mutable.



image.png


### 3.3.2. Operation

| Operation | Example | Description |
|------------|---------|-------------|
| Creating a dictionary | `crew_details={ "Pilot":"Kumar", "Co-pilot":"Raghav", "Head-Strewardess":"Malini", "Stewardess":"Mala" }` | First element in every pair is the key and the second element is the value. |
| Accessing the value using key | `crew_details["Pilot"]` | This will return the corresponding value for the specified key. |
| Iterating through the dictionary | `for key,value in crew_details.items():`<br>`     print(key,":",value)` | `items` function gives both key and value, which can be used in a for loop. |

### 3.3.3. Functions

| Function | Output | Explanation |
|----------|--------|-------------|
| `crew_details.get("Pilot")` | `Kumar` | Returns the value for given key. If the given key is not found, returns `None`. |
| `crew_details.update({"Flight Attendant":"Jane", "Co-pilot":"Henry"})` | No output, dictionary will be updated | Updates the dictionary with the given key-value pairs. If a key-value pair is already existing, it will be overwritten, otherwise it will be added to the dictionary. |

### 3.3.4. Example

- Creating a dictionary

````python
crew_details={
            "Pilot":"Kumar",
            "Co-Pilot":"Raghav",
            "Head-Strewardess":"Malini",
            "Stewardess":"Mala"
}
print(crew_details["Pilot"])

print("\nIterating the dictionary using items function")
for key,value in crew_details.items():
    print(key,":",value)
````

- Usually while working with dictionary, you will be interested in specific values.
- Let’s find the value of all pilots from crew_details.




> Note: Dictionary being unordered, the order of the values being displayed may vary during each execution of the above for loop.





````python
print("\nIterating the dictionary using keyword 'in'")
for key in crew_details:
    if(key=="Pilot" or key=="Co-Pilot"):
        print(crew_details[key])
````

- Dictionaries are mutable

````python
crew_details["Pilot"]="James" # Here the value for key "Pilot" is being updated to "James"
print("\nAfter modifying the value of Pilot:", crew_details["Pilot"])

print("------------------------------------------------------------------")
print("Before update:")
# Usage of get method()
print("Co-pilot:",crew_details.get("Co-Pilot"))

# Usage of update method()
crew_details.update({"Flight Attendant":"Jane", "Co-pilot":"Henry"})

print("\nAfter update:")
print("Co-pilot:",crew_details.get("Co-pilot"))
print("Flight Attendant:",crew_details["Flight Attendant"])
````

## 3.4. Set

## 6.1. Exception Handling: try & except & finally

Sometimes in programming we need to execute some code irrespective of whether the primary program logic itself succeeds or fails to do its job. In Python we can achieve this using a `finally` block.

A code written inside the `finally` block will **ALWAYS** be executed.

Default `except` block is the one without any type mentioned. It is the last `except` block.

```python
try:
    # Some Code....

except:
    # optional block
    # Handling of exception (if required)

else:
    # execute if no exception

finally:
    # Some code .....(always executed)
```

````python
balance=1000
amount="300Rs"

def take_card():
    print("Take the card out of ATM")

try:
    if balance>=int(amount):
        print("Withdraw")
    else:
        print("Invalid amount")

except TypeError:
    print("Type Error Occurred")
except ValueError:
    print("Value Error Occurred")
except:
    print("Some error Occurred")
finally:
    take_card()
````

| 3 | Actual |  | `encrypt_sentence` | `sentence-wel come` | N/A | `'lew cmoe'` | |

````python
def encrypt_sentence(sentence):
    #start writing your code here
    words = sentence.split(' ')
    encrypted_words = []

    vowels = "aeiouAEIOU"

    for i, word in enumerate(words):
        position = i + 1

        # Word at odd position: Reverse it
        if position % 2 != 0:
            encrypted_words.append(word[::-1])

        # Word at even position: Consonants before Vowels
        else:
            consonants = ""
            vowel_part = ""
            for char in word:
                if char in vowels:
                    vowel_part += char
                else:
                    consonants += char
            encrypted_words.append(consonants + vowel_part)

    return " ".join(encrypted_words)

sentence="The sun rises in the east"
encrypted_sentence=encrypt_sentence(sentence)
print(encrypted_sentence)
````

````python
def is_palindrome(word):
    # pass
    #Remove pass and write your logic here
    # Normalize the string to be case-insensitive
    word = word.lower()

    # Base Case: If the string has 0 or 1 characters, it is a palindrome
    if len(word) <= 1:
        return True

    # Check if the first and last characters are the same
    if word[0] == word[-1]:
        # Recursive Step: Check the inner substring excluding first and last characters
        return is_palindrome(word[1:-1])

    # If the characters don't match, it's not a palindrome
    return False

#Provide different values for word and test your program
result=is_palindrome("MadAMa")
if(result):
    print("The given word is a Palindrome")
else:
    print("The given word is not a Palindrome")
````
