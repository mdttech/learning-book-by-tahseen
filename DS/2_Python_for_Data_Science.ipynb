### 1.7.7. Sorting an Array

The NumPy array can be sorted by:

- Passing the array to the function `sort(array)`
- Using `array.sort()`

#### Difference Between `sort(array)` and `array.sort()`

Even though both functions are used for sorting, there is a difference between them.

- `array.sort()` modifies the original array by default.
- `sort(array)` does not modify the original array.

````python
#creating a list of 5 horsepower values
horsepower = [130, 165, 150, 150, 140]
#creating a numpy array from horsepower list
horsepower_arr = np.array(horsepower)
#using sort(array)
print('original array: ', horsepower_arr)
print('Sorted array: ', np.sort(horsepower_arr))
print('original array after sorting: ', horsepower_arr)
````

````python
#creating a list of 5 horsepower values
horsepower = [130, 165, 150, 150, 140]
#creating a numpy array from horsepower list
horsepower_arr = np.array(horsepower)
#using sort(array)
print('original array: ', horsepower_arr)
horsepower_arr.sort()
print('original array after sorting: ', horsepower_arr)
````

### 1.7.9 Broadcasting

"Broadcasting" refers to the term on how NumPy handles arrays with different shapes during arithmetic operations.

Array of smaller size is stretched or copied across the larger array.

#### Example

Consider the following arithmetic operations across 2 arrays.

In this example, `array2` is being stretched or copied to match `array1` during the arithmetic operation, resulting in a new array `array3` with the same shape as `array1`.



````python
import numpy as np
# Array 1
array1=np.array([5, 10, 15])
# Array 2
array2=np.array([5])
array3= array1 * array2
array3
````

#### Broadcasting Rules

The following diagram explains broadcasting.

image.png

---

#### Case 1

- Shape of first array = `1 × 3`
- Shape of second array = `1 × 1`

According to broadcasting rules, the second array gets stretched to match the shape of first array.

Resulting array shape = `1 × 3`

#### Case 2

- Shape of first array = `3 × 3`
- Shape of second array = `1 × 3`

According to broadcasting rules, the second array gets stretched to match the shape of first array.

Resulting array shape = `3 × 3`


#### Case 3

- Shape of first array = `3 × 1`
- Shape of second array = `1 × 3`

According to broadcasting rules, both first and second arrays get stretched.

Resulting array shape = `3 × 3`

Lee decides to walk `10000` steps every day to combat the effect that lockdown has had on his body’s agility, mobility, flexibility and strength.

Consider the following data from fitness tracker over a period of `10` days.

| Day Number | Steps Walked |
|---|---|
| 1 | 6012 |
| 2 | 7079 |
| 3 | 6886 |
| 4 | 7230 |
| 5 | 4598 |
| 6 | 5564 |
| 7 | 6971 |
| 8 | 7763 |
| 9 | 8032 |
| 10 | 9569 |

---

#### Tasks

i. Represent the above data in a `10 × 2` array.

   - First element in each row should contain day number.
   - Second element in each row should contain steps walked.


ii. Lee notices that the tracker’s battery dies every day at `7 pm`.

   Lee discovers that on an average, he walks `2000` steps every day after `7 pm`.

   Perform an appropriate operation on your array to add `2000` steps to all the observations.

iii. Write a program that returns the steps walked if the steps walked are more than `9000`.


iv. Print an array containing steps walked in sorted order.

````python
import numpy as np

# 1. Create the 10x2 array
data = np.array([
    [1, 6012],
    [2, 7079],
    [3, 6886],
    [4, 7230],
    [5, 4598],
    [6, 5564],
    [7, 6971],
    [8, 7763],
    [9, 8032],
    [10, 9569]
])

print("Original data:")
print(data)

# 2. Add 2000 steps to all observations in the second column
updated_data = data.copy()
updated_data[:, 1] = updated_data[:, 1] + 2000

print("\nData after adding 2000 steps:")
print(updated_data)

# 3. Return steps walked if more than 9000
steps_more_than_9000 = updated_data[updated_data[:, 1] > 9000, 1]

print("\nSteps greater than 9000:")
print(steps_more_than_9000)

# 4. Print sorted steps
sorted_steps = np.sort(updated_data[:, 1])

print("\nSorted steps:")
print(sorted_steps)
````

### 2.6.2. Masking Operation

The masking operation replaces values where the condition is `True`.

#### Student Marks Table

| Students | Chemistry | Physics | Mathematics | English |
|---|---|---|---|---|
| Subodh | 67 | 45 | 50 | 19 |
| Ram | 90 | 92 | 87 | 90 |
| Abdul | 66 | 72 | 81 | 72 |
| John | 32 | 40 | 12 | 68 |

#### Problem Statement

The teacher does not want to reveal the marks of students who have failed.

Condition:

- If marks `>= 33` → Passed
- Otherwise → Failed

The marks of failed students have to be replaced with `"Fail"`.

---

#### Solution

First, create the DataFrame containing student marks.

---

##### Syntax

`DataFrame.mask(cond, other = nan, inplace = False, axis = None)`

---

##### Parameters

- `cond`
  - Where condition is `False`, original value is kept.
  - Where condition is `True`, value is replaced.

- `other`
  - Values used to replace entries where condition is `True`.

- `inplace`
  - Whether to perform the operation directly on the original DataFrame.

- `axis`
  - Alignment axis.


##### Example

A condition is created where marks less than `33` are identified.

The `mask()` function is then used to replace those values with `"Fail"`.

````python
marks = [{'Chemistry': 67, 'Physics': 45, 'Mathematics': 50, 'English' : 19},
        {'Chemistry': 90, 'Physics': 92, 'Mathematics': 87, 'English' : 90},
        {'Chemistry': 66, 'Physics': 72, 'Mathematics': 81, 'English' : 72},
        {'Chemistry': 32, 'Physics': 40, 'Mathematics': 12, 'English' : 68}]
marks_df = pd.DataFrame(marks, index = ['Subodh', 'Ram', 'Abdul', 'John'])
marks_df
````

````python
f = marks_df < 33
marks_df.mask(f, 'Fail')
````

#### Problem Statement 1

#### Problem Statement 2

Senior engineers at XYZ Custom Cars want to understand the effect of:

- Model year
- Number of cylinders

on horsepower.

#### Solution

One approach is to check:

- Mean horsepower
- Minimum horsepower
- Maximum horsepower

based on:

- Number of cylinders
- Model year

For this requirement, the `agg()` function is combined with `groupby()`.

##### Steps Performed

1. Group the DataFrame using:
   - `cylinders`
   - `model_year`

2. Apply aggregation functions on `horsepower`:
   - `mean`
   - `min`
   - `max`

3. Rename grouped DataFrame columns:
   - `hp_mean`
   - `hp_min`
   - `hp_max`

4. Reset the index.

5. View the first few rows using `head()`.



````python
#Creating a DataFrame grouped on cylinders and model_year and finding mean, min and max of horsepower
grouped_multiple = df.groupby(['cylinders', 'model_year']).agg({'horsepower': ['mean', 'min', 'max']})
#Naming columns in grouped DataFrame
grouped_multiple.columns = ['hp_mean', 'hp_min', 'hp_max']
#Resetting index
grouped_multiple = grouped_multiple.reset_index()
#Viewing head of resulting DataFrame
grouped_multiple.head()
````

## 2.8. Exercise

The rainfall dataset to be considered for this exercise contains region (district) wise rainfall across India.

### Tasks

1. Import the data into Python environment as a Pandas DataFrame.

2. Check for missing values, if any, and drop the corresponding rows.

3. Find the district that gets the highest annual rainfall.

4. Display the top `5` states that get the highest annual rainfall.

5. Drop the following columns:

- `Jan-Feb`
- `Mar-May`
- `Jun-Sep`
- `Oct-Dec`

6. Display the state-wise mean rainfall for all the months using a pivot table.

7. Display the count of districts in each state.

8. For each state, display:

- The district that gets the highest rainfall in May
- The recorded rainfall

````python
import pandas as pd

# Load dataset
rainfall_df = pd.read_csv('rainfall.csv')

# Remove missing values
rainfall_df = rainfall_df.dropna()

# District with highest annual rainfall
highest_rainfall_district = rainfall_df.loc[rainfall_df['ANNUAL'].idxmax()]
print("\nDistrict with Highest Annual Rainfall")
print(highest_rainfall_district[['DISTRICT', 'STATE_UT_NAME', 'ANNUAL']])

# Top 5 states with highest annual rainfall
state_annual_rainfall = rainfall_df.groupby('STATE_UT_NAME')['ANNUAL'].sum()
print("\nTop 5 States with Highest Annual Rainfall")
print(state_annual_rainfall.sort_values(ascending=False).head(5))

# Drop unwanted columns
rainfall_df = rainfall_df.drop(columns=['Jan-Feb', 'Mar-May', 'Jun-Sep', 'Oct-Dec'])

# Pivot table
pivot_table = pd.pivot_table(
    rainfall_df,
    values=['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN',
            'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC'],
    index='STATE_UT_NAME',
    aggfunc='mean'
)

print("\nState-wise Mean Rainfall")
print(pivot_table)

# District count in each state
print("\nDistrict Count in Each State")
print(rainfall_df.groupby('STATE_UT_NAME')['DISTRICT'].count())

# Highest May rainfall district in each state
max_may_idx = rainfall_df.groupby('STATE_UT_NAME')['MAY'].idxmax()
max_may_rainfall = rainfall_df.loc[max_may_idx, ['STATE_UT_NAME', 'DISTRICT', 'MAY']]

print("\nHighest Rainfall in May for Each State")
print(max_may_rainfall)
````

Scikit-learn (also referred as `sklearn`) is a Python library widely used for machine learning.

## 4.2. Problem Statement

Engineers at XYZ custom cars want to create a machine learning model that can predict the `mpg` of any car that comes to their garage.

`MPG` refers to **Miles per Gallon**.

### Solution:

A **Linear Regression** model is to be made for this problem.

The different stages followed in ML model building are shown below.

## 4.3. Loading the Data

Let us start with loading the data.

````python
import pandas as pd
df = pd.read_csv('auto_mpg.csv')
df
````

## 4.4. Data Preprocessing

### 4.4.1. Data Properties

`DataFrame.info()` is used to get information about the data properties.



````python
df.info()
````

### 4.4.2. Dropping Null Values

It can be observed that the `horsepower` attribute has some null values.

The easiest approach is to remove the rows with null values.


````python
df.dropna(inplace = True)
df.info()
````

### 4.4.3. Predictors and Target

The target variable is `mpg` which has to be predicted.

The predictors are the variables that are used to predict the target.

Here, except the name of the car, all the other variables are included as predictors.

````python
#Creating matrix of predictors
X = df.iloc[:, 1:8]
#Creating target
y = df.iloc[:, 0]
````

### 4.4.4. Encoding Categorical Variables

Since the `origin` feature is a categorical variable, `get_dummies()` function can be used from Pandas to encode it.

The categorical variable `origin` gets encoded with `0s` and `1s`.

````python
X = pd.get_dummies(X)
X
````

## 4.5. Train and Test Split

The data must be divided into two parts:

- Training set → Used to train the model
- Testing set → Used to validate the model

The `sklearn` library is used for this purpose.

### Important Parameters

- `X_train` and `y_train` correspond to training predictors and target respectively.
- `X_test` and `y_test` correspond to testing predictors and target respectively.
- `test_size = 0.2` represents that `20%` of the data will be used as test set.

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
````

## 4.6. Applying Standard Scaler on the Data

Since all the variables in the data are with different units of measurements and different scales, it is a good idea to standardize them.

A Standard Scaler transforms the columns such that:

- Mean of every column becomes `0`
- Standard deviation becomes `1`

````python
#Applying standard scaler on the data
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
scale.fit_transform(X_train)
scale.transform(X_test);
````



## 4.7. Linear Regressor

The Linear Regression model is used to build the model.

A linear regression model uses the following equation:

$$
y = B_0 + B_1X_1 + B_2X_2 + \dots + B_nX_n
$$

Where:

- \( y \) → Target variable
- \( X_1, X_2, \dots, X_n \) → Predictors
- \( B_0 \) → Intercept
- \( B_1, B_2, \dots, B_n \) → Coefficients

### 4.7.1. Building the Model

The Linear Regression model is fitted on the training data set using `sklearn`.

After fitting:

- `m` represents the coefficient (slope)
- `c` represents the intercept




````python
#Importing and fitting the model on training set
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
#Fitting the model on training data :
reg.fit(X_train, y_train)
#Checking the coefficient(slope) and intercept.
#'m' represents the coefficient and 'c' represents the intercept.
m = reg.coef_
c = reg.intercept_
m,c
````

### 4.7.2. Prediction

The trained model is used for prediction against:

- Training data set
- Testing data set

Predicted values are stored in:

- `y_pred_train`
- `y_pred_test`

````python
#Predicting the target: mpg against the predictors in the training data set
#Predicted data stored in y_pred_train
y_pred_train = reg.predict(X_train)
#Predicting the target: mpg against the predictors in the testing data set
#Predicted data stored in y_pred_test
y_pred_test = reg.predict(X_test)
````

## 4.8. Model Evaluation

Different metrics are used to evaluate the performance of the model.

Here, the **R-Squared Score** is used.

### 4.8.1. Training Accuracy

The R-squared score measures how close the predicted value of `mpg` is to the real value in the training data set.

The training accuracy comes out to be:

$$
81\%
$$



````python
# Prediction Accuracy in terms of how close is the predicted value of target: mpg
# to the real value in training data set
from sklearn.metrics import r2_score
r2_S = r2_score(y_train, y_pred_train)
r2_S
````

### 4.8.2. Testing Accuracy

The R-squared score is also calculated on the testing data set.

The testing accuracy comes out to be:

$$
83\%
$$

````python
# Prediction Accuracy in terms of how close the predicted value of target: mpg
# to the real value in testing data set
from sklearn.metrics import r2_score
r2_S = r2_score(y_test, y_pred_test)
r2_S
````
