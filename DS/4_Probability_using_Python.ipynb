# 1. Introduuction

## 1.1. Chance

Chance is a possibility of something happening.


| Situation | Chance |
|---|---|
| Black colour sun | Impossible (0% chance) |
| Monday followed by Tuesday | Certain (100% chance) |
| Having pizza for dinner | Uncertain (some chance) |

When a chance takes us to a situation where the outcome cannot be predicted exactly (i.e. there is an unclear possibility of occurrence), Probability is used to quantify the variability in the outcome.



## 1.2. Probability

By Probability, we measure the likelihood of an event in which we are interested.

## 1.3. Likelihood of an Event

The likelihood of an event is the frequency with which the event may occur.



## 1.4. Probability Use

### 1.4.1. Describe the Frequency of Occurrence

When we want to know the chance of occurrence of an outcome based on the number of times it has occurred, probability is used to calculate the frequency with which the event has occurred.

#### Example

Suppose a coin is tossed 10 times and results in:

- 6 Heads
- 4 Tails

We may want to calculate the chance that a head appears in the next toss.

#### Inference

- The coin was tossed repeatedly.
- Previous outcomes help estimate the likelihood of getting a head in the subsequent toss.

### 1.4.2. Describe a Subjective Belief

Probability can also be used to understand uncertainty associated with a belief.

#### Example

A scholar asserted that Poem A and Poem B were written by the same poet with a 95% belief.

#### Inference

- The assertion conveys a belief rather than frequency.
- The event is a one-time event and not a repeated experiment.


## 1.5. Frequency of Occurrence

Consider an activity (experiment) of tossing a coin 10 times. This activity results in heads and tails (outcomes).

How can we proceed when we want to calculate "the chance of getting head more than the number of times the tail has appeared."

In order to answer questions of this kind, it is essential to understand the frequency with which each value (outcome) occurred.

The activity (experiment), results of activity (outcomes) and the frequency/likelihood with which the values (outcomes) may occur can be described using a Probabilistic Model.

A probabilistic model is a generic structure that describes the random outcomes of an activity.

# 3. Counting Techniques

## 3.1. Introduction

In a random experiment, it is often necessary to count the number of possible outcomes.

The simplest way to determine the number of outcomes is to list all possible outcomes of the experiment.

However, in many real-world problems, the number of outcomes can be very large, making it impractical or impossible to list them all.

For such situations, specialized counting techniques are used to determine the number of possible outcomes efficiently.


### Common Counting Techniques

1. **Product Rule**
2. **Sum Rule**
3. **Combinations**
4. **Permutations**

## 3.2. Product Rule

The Product Rule is used to count the total number of outcomes in a multi-stage experiment.

If an experiment consists of multiple stages, the total number of ways to complete the experiment is the product of the number of choices available at each stage.

### Example 1

Consider a 3-stage experiment:

image.png


- Stage A has 4 choices
- Stage B has 2 choices
- Stage C has 3 choices

### Total Number of Ways

$$
4 \times 2 \times 3 = 24
$$

That is,

$$
(\text{Choices in Stage A})
\times
(\text{Choices in Stage B})
\times
(\text{Choices in Stage C})
$$

## Example 2

### Scenario

Consider stocks from two sectors:



#### Banking Sector

$$
\text{Stocks(Bank)} = \{\text{Bank1, Bank2, Bank3}\}
$$

#### Automotive Sector

$$
\text{Stocks(Auto)} = \{\text{Auto1, Auto2, Auto3, Auto4}\}
$$

An investor wants to invest in:

- One banking stock **AND**
- One automotive stock

### Total Number of Investment Choices


image.png



$$
3 \times 4 = 12
$$

That is,

$$
(\text{Number of Bank Stocks})
\times
(\text{Number of Automotive Stocks})
$$

Therefore, there are **12 different ways** to make the investment.

## 3.3. Sum Rule

### Scenario

Consider stocks in the banking and automotive sectors.

If an investor wants to invest in either banking **OR** automotive stocks, how many total possible ways can the investor invest?

## 3.4. Combination

### Scenario

Consider stocks of the automotive sector.

$$
\text{Stocks(Auto)} = \{\text{Auto1, Auto2, Auto3, Auto4}\}
$$

If an investor wants to invest in 2 automotive stocks, in how many possible ways can the investor invest?

### Total Number of Ways

$$
^nC_r
=
\frac{n!}{r!(n-r)!}
$$

Where:

- $n$ = Total number of automotive stocks
- $r$ = Number of stocks the investor wants to invest in

For this example:

$$
{}^4C_2
=
\frac{4!}{2!(4-2)!}
=
6
$$

Therefore, the total number of possible ways is **6**.

### Combination Without Replacement

#### Scenario

Suppose an urn contains:

- 3 red balls
- 4 green balls

We want to calculate the number of ways to select:

- 2 red balls
- 2 green balls

> **Note:** Replacement is not allowed.

#### Solution

Ways to choose 2 red balls from 3 red balls:

$$
{}^3C_2
=
\frac{3!}{2!(3-2)!}
=
3
$$

Ways to choose 2 green balls from 4 green balls:

$$
{}^4C_2
=
\frac{4!}{2!(4-2)!}
=
6
$$

Total ways:

$$
^3C_2 \times ^4C_2
$$

$$
= 3 \times 6
$$

$$
= 18
$$

### Combination With Replacement

#### Scenario

An urn contains:

- 3 identical red balls
- 4 identical green balls

We want to choose:

- 2 red balls
- 2 green balls

> **Note:** Replacement is allowed.

#### Solution

Ways to choose:

- First red ball:
$$
{}^3C_1
=
\frac{3!}{1!(3-1)!}
=
3
$$
- Second red ball:
$$
{}^3C_1
=
\frac{3!}{1!(3-1)!}
=
3
$$
- First green ball:

$$
{}^4C_1
=
\frac{4!}{1!(4-1)!}
=
4
$$
- Second green ball:

$$
{}^4C_1
=
\frac{4!}{1!(4-1)!}
=
4
$$

Total ways:

$$
^3C_1 \times ^3C_1 \times ^4C_1 \times ^4C_1
$$

$$
= 3 \times 3 \times 4 \times 4
$$

$$
= 144
$$

# 4. Probability

Consider a random experiment with a finite sample space $S$.

Assignment of a real number to each event of the sample space is known as a **probability measure** that depicts the likelihood of occurrence of the event.

Suppose a random experiment is repeated $n$ times.

If Event $A$ occurs $n(A)$ times, then the probability of Event $A$ is given by:

$$
P(A)=\frac{n(A)}{n}
$$

Where:

- $P(A)$ is a real number
- $n > 0$
- $n(A)$ is the number of times Event $A$ has occurred



## Axioms of Probability

For a finite sample space $S$, consider an Event $A$.

The probability $P(A)$ assigned to Event $A$ must satisfy the following axioms:

1. $P(A) \geq 0$

2. $P(S) = 1$

3. If Event $A$ and Event $B$ are mutually exclusive, then:

$$
P(A \cup B) = P(A) + P(B)
$$

and

$$
A \cap B = \varnothing
$$



## Properties of Probability

Using the axioms, the following properties can be obtained:

$$
P(A^c) = 1 - P(A)
$$

$$
P(\varnothing) = 0
$$

$$
P(A) < P(B)
\quad \text{if} \quad
A \subset B
$$

$$
P(A) \leq 1
$$

$$
P(A \cup B)
=
P(A)+P(B)-P(A \cap B)
$$



### Scenario 1

Consider a box containing:

- 3 red balls
- 2 blue balls
- 1 green ball

One ball is selected randomly.

What is the probability that the selected ball is blue?

### Solution

Total possible outcomes:

$$
3+2+1=6
$$

Total ways a blue ball can be drawn:

$$
{}^2C_1 = 2
$$

Probability of drawing a blue ball:

$$
P(\text{Blue})
=
\frac{2}{6}
=
\frac{1}{3}
$$



### Scenario 2

A fair die is rolled 12 times.

The number 4 appears 5 times.

What is the probability of obtaining the number 4 for the given experiment?

### Solution

Total possible outcomes:

$$
12
$$

Number of times 4 appeared:

$$
5
$$

Probability of rolling 4:

$$
P(4)
=
\frac{5}{12}
$$

# 5. Conditional Probability

Conditional Probability is used to determine the likelihood of an event when partial information about the event is known.



## Examples

### Example 1

In an experiment of rolling a die twice, it was observed that the sum of the outcomes was even.

What is the chance that the first outcome was 2?

### Example 2

In an experiment where a word is to be guessed, it was observed that the first letter of the word is 'A'.

What is the chance that the second letter is 'b'?



### Observation

In the above examples, partial information about the outcome of the experiment is available.

- In Example 1, the sum is known to be even.
- In Example 2, the first letter is known.



## Conditional Probability

### Scenario

In an experiment of rolling a die, what is the likelihood that the outcome is 2, given that the outcome is even?

Consider a fair roll of a die, i.e., all outcomes have equal probability:

$$
P(\text{Outcome})=\frac{1}{6}
$$

Let:

- Event $A$ = Outcome is 2
- Event $B$ = Outcome is even

Then:

$$
P(A|B)
$$

represents

$$
P(\text{Outcome is 2} \mid \text{Outcome is even})
$$



### Formula

$$
P(A|B)
=
\frac{\text{Number of elements of }(A \cap B)}
{\text{Number of elements of }B}
$$

or equivalently,

$$
P(A|B)
=
\frac{P(A \cap B)}
{P(B)}
$$



### Solution

Since the outcome is even (Event $B$), the possible outcomes are:

$$
\{2,4,6\}
$$

Therefore,

$$
n(B)=3
$$

Event $A$ contains only one favorable outcome:

$$
\{2\}
$$

Therefore,

$$
n(A \cap B)=1
$$

Hence,

$$
P(A|B)
=
\frac{n(A \cap B)}
{n(B)}
=
\frac{1}{3}
$$

or

$$
P(A|B)
=
\frac{P(A \cap B)}
{P(B)}
=
\frac{1}{3}
$$



> **Note:** Conditional probability is undefined if the conditioning event has zero probability. Therefore, $P(B) > 0$ is mandatory.

# 6. Total Probability and Bayes' Theorem

### Scenario

A retailer procures bread from three suppliers:

- $40\%$ from $S_1$
- $35\%$ from $S_2$
- $25\%$ from $S_3$

It was found that:

- $5\%$ of bread supplied by $S_1$ is defective
- $2\%$ of bread supplied by $S_2$ is defective
- $3\%$ of bread supplied by $S_3$ is defective

Assumptions:

- No other supplier supplies bread (**Exhaustive Events**)
- All suppliers are different (**Mutually Exclusive Events**)



image.png


### Questions

1. What is the probability that a customer picks a defective bread?

2. What is the probability that a selected defective bread was supplied by $S_2$?


image.png




## Probability of Selecting a Defective Bread

Let:

$$
D = \text{Event that a defective bread is selected}
$$

Then,

$$
P(D)
=
P(S_1)P(D|S_1)
+
P(S_2)P(D|S_2)
+
P(S_3)P(D|S_3)
$$

Substituting the values:

$$
P(D)
=
(0.40 \times 0.05)
+
(0.35 \times 0.02)
+
(0.25 \times 0.03)
$$

$$
=
0.0200 + 0.0070 + 0.0075
$$

$$
=
0.0345
$$

Therefore,

$$
P(D)=0.0345
$$



## Law of Total Probability

For $n$ mutually exclusive and exhaustive events:

$$
P(D)
=
\sum_{i=1}^{n}
P(S_i)P(D|S_i)
$$



## Probability that a Defective Bread was Supplied by $S_2$

We need to calculate:

$$
P(S_2|D)
$$

Using Bayes' Theorem:

$$
P(S_2|D)
=
\frac{P(S_2)P(D|S_2)}
{
P(S_1)P(D|S_1)
+
P(S_2)P(D|S_2)
+
P(S_3)P(D|S_3)
}
$$

Substituting the values:

$$
P(S_2|D)
=
\frac{0.35 \times 0.02}
{
(0.40 \times 0.05)
+
(0.35 \times 0.02)
+
(0.25 \times 0.03)
}
$$

$$
=
\frac{0.007}{0.0345}
$$

$$
\approx 0.2029
$$

Therefore,

$$
P(S_2|D)\approx 0.2029
$$



## General Form

The probability that a defective bread was supplied by supplier $k$ among $n$ suppliers is:

$$
P(S_k|D)
=
\frac{P(S_k)P(D|S_k)}
{
\sum_{i=1}^{n}
P(S_i)P(D|S_i)
}
$$



## Multiplication Rule of Probability

For Events $A$ and $B$:

If $P(A)\neq0$,

$$
P(A \cap B)
=
P(A)P(B|A)
$$

If $P(B)\neq0$,

$$
P(A \cap B)
=
P(B)P(A|B)
$$



## Bayes' Theorem

Using

$$
P(A \cap B)
=
P(A)P(B|A)
=
P(B)P(A|B)
$$

we obtain:

$$
P(A|B)
=
\frac{P(A)P(B|A)}
{P(B)}
$$

# 7. Random Variable
## 7.1. Scenario

Let us consider a scenario to understand a random variable.

Sashi, an ice-cream seller, wants to sell home-made ice-creams and waffle-cones using a mobile ice-cream van.

She plans to sell five different flavors:

- Vanilla
- Chocolate
- Strawberry
- Black-Currant
- Pistachio



Sashi plans to begin selling ice-creams in the month of March.

She manufactures a batch of 1200 waffle-cones and examines random waffle-cones, judging each cone as either:

- Defective (D)
- Non-Defective (N)

She decides to examine 3 waffle-cones.



### Experiment

The process of observation of an activity is termed an **experiment**.

The results of the observation are termed **outcomes** of the experiment.

Random experiments are those whose outcomes cannot be predicted exactly.

In this scenario:

- Examining a random waffle-cone and judging it as defective or non-defective is the experiment.
- The possible outcomes are:
  
$$
\{D,\;N\}
$$

Since the exact outcome cannot be predicted, this is a **random experiment**.



### Trial

Individual repetitions of the same random experiment are termed **trials**.

Examining a single waffle-cone is one trial within the experiment consisting of examining 3 waffle-cones.



### Sample Space

The set of all possible outcomes of a random experiment is called the **sample space**.

For 3 trials, the sample space is:

$$
S=
\{
DDD,\;
DDN,\;
DND,\;
DNN,\;
NDD,\;
NDN,\;
NND,\;
NNN
\}
$$


image.png


### Event

One particular outcome or a set of outcomes from the sample space is termed an **event**.

#### Example 1

Event of interest: Total number of defective cones is exactly 2.

$$
E=
\{
DDN,\;
DND,\;
NDD
\}
$$

Therefore, there are **3 ways** in which this event can occur.



#### Example 2

Event of interest: Total number of defective cones is one or more.

$$
E=
\{
DDD,\;
DNN,\;
NDN,\;
NND,\;
DDN,\;
DND,\;
NDD
\}
$$

Therefore, there are **7 ways** in which this event can occur.

## 7.2. Random Variable

A function $X$ can be defined on the sample space such that each outcome is mapped to a numerical value based on the event of interest.

For example, let

$$
X=
\text{Total number of defective cones in 3 trials}
$$

Then:


image.png






### Definition

A real-valued function defined over a sample space is called a **Random Variable**.

Properties:

- Only one real value is assigned to each outcome.
- $X$ (or any uppercase letter) denotes a random variable.
- A lowercase letter, such as $x$, denotes a particular value taken by the random variable.



### Important Note

$X$ is **not** a variable as used in Algebra.

Example of an algebraic variable:

$$
x+2=7
$$

where the value of $x$ is unknown.

A random variable is a **function**.

It can be represented as:

$$
X=\{x_1,x_2,x_3,\ldots\}
$$

or

$$
X=x_k,
\quad k=1,2,3,\ldots
$$



### Random Variable in the Ice-Cream Scenario

Let

$$
X=
\text{Total number of defective cones in 3 trials}
$$

Then,

$$
X=\{0,1,2,3\}
$$



### Observations

- The possible values of $X$ are known before performing the experiment.
- However, the actual value that $X$ takes cannot be predicted before the experiment.
- $X$ can take different values each time the experiment is performed.

Due to its trial-to-trial variability and unpredictable nature, $X$ is called a **Random Variable**.

## 7.3. Types of Random Variables

A random variable may either be:

- Discrete Random Variable
- Continuous Random Variable



### 7.3.1. Discrete Random Variable

A random variable having a finite range is called a **Discrete Random Variable**.

#### Examples

- Total number of scratches found on a glass surface
- Proportion or percentage of defective boxes among 1000 tested

In real-time experiments, the total number of transmitted bits in error can be recorded as:

- Integers
- Fractions such as $0.0042$ of 10,000 transmitted bits

Whenever the count is limited to finite points on the real number line, the random variable is called a discrete random variable.



#### Example from the Ice-Cream Scenario

Possible outcomes for:

$$
X = \text{Number of defective cones in 3 trials}
$$

are

$$
X=\{0,1,2,3\}
$$

$X$ is a discrete random variable because it assumes a finite set of values.



#### Other Examples from the Ice-Cream Scenario

- Number of customers expected to arrive between 2 PM and 3 PM
- Number of ice-creams bought by children below 12 years of age
- Number of ice-cream cones that break/leak while being filled
- Number of ice-creams sold in a month
- Number of ice-creams sold in a year



### 7.3.2. Continuous Random Variable

A random variable having values within a real-number interval is called a **Continuous Random Variable**.

#### Examples

- Height of a child measured between ages 2 and 6
- Weight of a person recorded over a span of 2 years

Continuous random variables can assume values over a range of real numbers.

Although values are often rounded in practical applications, theoretically they can take infinitely many values within an interval.



#### Note

## 9.1. Probability Mass Function (PMF)

The Probability Mass Function (PMF), also known as the mass of a discrete random variable, defines that the probability of a random variable with a discrete value is exactly equal to some value.

Let us assume a discrete random variable $X$ having a finite range of outcomes or countably infinite possible outcomes.

$$
X=\{x_1,x_2,x_3,\ldots\}
$$

or

$$
X=x_k
\quad \text{where } k=1,2,3,\ldots
$$

The PMF $f$ of a discrete random variable $X$ is given by:

$$
f(x_k)=P(X=x_k)
$$

where:

- $k=1,2,3,\ldots$
- $P(X=x_k)$ is the probability of occurrence of the event $X=x_k$


### Properties of PMF

1. Non-Negativity

$$
f(x_k)\geq 0
$$

for all real values of $x_k$ in the given range.

2. Total Probability

$$
\sum f(x_k)=1
$$

## 9.2. Cumulative Distribution Function (CDF)

The probability given by a function of having a value less than or equal to the argument of the function for a random variable is called the **Cumulative Distribution Function (CDF)**.

For a random variable $X$, the CDF is defined as:

$$
F(x)=P(X\leq x)
$$

for all real numbers $x$, where $P$ denotes the probability of occurrence of the event $X\leq x$.



### Important Note

CDF can describe any type of probability distribution.

It can be:

- Discrete
- Continuous



### Properties of CDF

For all $a \leq b$,

$$
P(X\leq b)
=
P(X\leq a)
+
P(a<X\leq b)
$$

Therefore,

$$
F(b)
=
F(a)
+
P(a<X\leq b)
$$

and

$$
P(a<X\leq b)
=
F(b)-F(a)
$$



### Relationship Between PMF and CDF

If the PMF of a discrete random variable $X$ is known, then its CDF can be calculated.

If the range of $X$ is

$$
X=\{x_1,x_2,x_3,\ldots\}
$$

then

$$
F(x)
=
\sum_{x_k \leq x}
P(X=x_k)
$$

where $P$ denotes the probability of occurrence.



### Finding $P(X<x)$

CDF gives:

$$
P(X\leq x)
$$

For a discrete random variable,

$$
P(X<x)
=
P(X\leq x)-P(X=x)
$$

Therefore,

$$
P(X<x)
=
F(x)-f(x)
$$

where:

- $F(x)$ = CDF
- $f(x)$ = PMF

# 9.b. The Binomial Distribution

The characteristics of a Binomial Random Experiment are:

- Fixed number of trials
- A success or a failure is the outcome of a single trial
- All trials are independent of each other
- All trials are identical and have the same probability of occurrence

Trials satisfying the above four conditions are called **Bernoulli Trials**.


## Binomial Random Variable

A random variable $X$ that equals the number of trials resulting in success ($x$) among $n$ Bernoulli trials is called a **Binomial Random Variable** with parameters:

$$
0<p<1
$$

and

$$
n=1,2,3,\ldots
$$



### Definitions

Let:

- $n$ = Total number of trials
- $x$ = Number of successful trials
- $n-x$ = Number of failed trials
- $p$ = Probability of success
- $q=1-p$ = Probability of failure

The probability of obtaining $x$ successes in $n$ trials is described by the **Binomial Distribution**.



## Developing the Binomial Distribution

To develop the probability distribution for a binomial random experiment:

1. Determine the probability of any one way in which the event of interest (success $x$) can occur.
2. Multiply this probability by the total number of ways in which the event can occur.



### Number of Possible Ways

The total number of possible ways is given by:


$$
C(n,x)
=
{}^{n}C_x
=
\frac{n!}{x!(n-x)!}
$$



where:

$C(n,x)={}^{n}C_x$ = Number of combinations of $x$ objects selected from $n$ objects,

where

$$
n \geq x
$$

Also,

$$
n!=n(n-1)(n-2)\cdots(2)(1)
$$

and

$$
0!=1
$$

(by definition)



### Probability of One Particular Arrangement

The probability of one particular way in which the event can occur is:

$$
p^xq^{\,n-x}
$$



## Probability Mass Function (PMF)

The PMF of a Binomial Random Variable $X$ is:

$$
f(x)
=
P(X=x)
=
C(n,x)p^x(1-p)^{n-x}
$$

or equivalently,

$$
f(x)
=
P(X=x)
=
C(n,x)p^xq^{\,n-x}
$$

where:

- $C(n,x)$ = Binomial Coefficient
- $n$ = Total number of Bernoulli trials
- $x=0,1,\ldots,n$ = Number of successes
- $n-x$ = Number of failures
- $p$ = Probability of success
- $q=1-p$ = Probability of failure



## SciPy Functions for Binomial Distribution

The following functions are available in the `scipy.stats` module:

### PMF

$$
\texttt{binom.pmf(x,\;n,\;p)}
$$

### CDF

$$
\texttt{binom.cdf(x,\;n,\;p)}
$$

### Quantile Function

$$
\texttt{binom.ppf(q,\;n,\;p)}
$$

The smallest value $x$ such that

$$
F(x)\geq p
$$

is called the **quantile**, where $F$ is the CDF.

### Random Generation

$$
\texttt{binom.rvs(n,\;size=10,\;p)}
$$

where:

- $x$ = Number of successes
- $n$ = Number of fixed trials
- $p$ = Probability of success
- $q$ = Binomial CDF of $x$
- `size` = Target number of generated observations

# 9.c. The Negative Binomial Distribution

The PMF of a Negative Binomial Random Variable $X$ is:

$$
f(x)
=
P(X=x)
=
C(x-1,r-1)p^r(1-p)^{x-r}
$$

or equivalently,

$$
f(x)
=
P(X=x)
=
C(x-1,r-1)p^rq^{\,x-r}
$$

where:

$$
x=r,r+1,r+2,\ldots
$$

and

- $p$ = Probability of success
- $q=1-p$ = Probability of failure



## Range of the Random Variable

The experiment must be performed at least $r$ times to obtain $r$ successes.

Therefore, the range of $X$ is:

$$
X=\{r,r+1,r+2,\ldots\}
$$

or

$$
r \leq X < \infty
$$



## Relation with Binomial Distribution

For a Binomial Random Variable:

- Number of trials is fixed
- Number of successes is random

For a Negative Binomial Random Variable:

- Number of successes is fixed
- Number of trials is random

Hence, the Negative Binomial Distribution can be viewed as the opposite (or negative) of the Binomial Distribution.



## SciPy Functions for Negative Binomial Distribution

### Quantile Function

$$
\texttt{nbinom.ppf(q,\;n,\;p)}
$$

### Random Generation

$$
\texttt{nbinom.rvs(n,\;p,\;size)}
$$

where:

- $x$ = Number of trials required to achieve $n$ successes
- $n$ = Number of successes
- $p$ = Probability of success
- $q$ = Negative Binomial CDF of $x$
- `size` = Target number of generated observations



## Parameters

For a Negative Binomial Random Variable:

- $n$ = Fixed number of successes

# 9.d. The Poisson Distribution

Consider an interval of real numbers and assume that counts (successes or outcomes of interest) occur randomly throughout the interval.

The interval can be partitioned into sufficiently small sub-intervals such that each sub-interval satisfies the following properties:

1. The probability of more than one success is almost zero.

$$
P(\text{success}>1)\approx 0
$$

2. The probability of exactly one success is the same for every sub-interval and is proportional to the length of the sub-interval.

$$
P(\text{success}=1)\propto \text{Sub-Interval Length}
$$

3. The number of successes in each sub-interval is independent of the number of successes in other sub-intervals.

If a random experiment satisfies these three conditions, it is called a **Poisson Process**.

> Historically, the term *process* has been used to indicate observation of a system over a period of time.



## Poisson Random Variable

A random variable $X$ that equals the number of counts in the interval of a Poisson Process is called a **Poisson Random Variable** with parameter:

$$
\lambda > 0
$$

The PMF of $X$ is:


$$
f(x)
=
P(X=x)
=
\frac{e^{-\lambda}\lambda^x}{x!}
$$


where:

- $x=0,1,2,3,\ldots$ = Number of counts (total successes) in the interval
- $\lambda$ = Expected number of successes in the interval
- $e=2.71828$ = Euler's number



## SciPy Functions for Poisson Distribution

### CDF

$$

### Quantile Function

$$
\texttt{poisson.ppf(q,mu)}
$$

### Random Generation

$$
\texttt{poisson.rvs(mu,size)}
$$

where:

- $x$ = Number of successes
- $mu$ = $\lambda$
- $q$ = Poisson CDF of $x$
- `size` = Target number of generated observations

# 10. Continuous Distributions

A Continuous Distribution is a probability distribution in which the random variable can take an infinite number of possible values within a given interval.





## 10.1. Probability Density Function (PDF)

For Continuous Random Variables, the PMF does not work because:

$$
P(X=x)=0
$$

for every value of $x$ on the real number line.

Since the probability at a single point is zero, we define the **Probability Density Function (PDF)** instead of the PMF.

The PDF gives the **density of probability** rather than the probability mass.

The Probability Density Function of a Continuous Random Variable $X$ is defined as:

$$
f(x)
=
\lim_{\Delta \to 0}
\frac{P(x<X\leq x+\Delta)}{\Delta}
$$

wherever the limit exists.

The function $f(x)$ gives the probability density at the point $x$.

It is the limit of:

$$
\frac{\text{Interval Probability}}{\text{Interval Length}}
$$

as

$$
\Delta \rightarrow 0
$$



You also know that:

$$
P(x < X \leq x+\Delta)
=
F(x+\Delta)-F(x)
$$

Therefore,

$$
f(x)
=
\lim_{\Delta \to 0^{+}}
\frac{F(x+\Delta)-F(x)}{\Delta}
=
\frac{dF(x)}{dx}
=
F'(x)
$$

where:

- $F(x)$ is the Cumulative Distribution Function (CDF)
- $F'(x)$ denotes the derivative of $F(x)$



### Interpretation of PDF

The Probability Density Function describes the relative likelihood of a continuous random variable taking values in an interval.

For a continuous random variable $X$ having an uncountably infinite number of possible outcomes:

$$
f(x)
=
F'(x)
=
\frac{d}{dx}F(x)
$$

The probability that $X$ falls within the infinitesimal interval:

$$
[x,\;x+dx]
$$

is:

$$
f(x)\,dx
$$



### Properties of PDF

1. Non-Negativity

$$
f(x)\geq 0
$$

for all real values of $x$.

2. Total Probability Equals One

$$
\int_{-\infty}^{\infty} f(x)\,dx = 1
$$



### Observation

The Probability Density Function (PDF) is used to describe **Continuous Probability Distributions**.


image.png





## 10.2. Cumulative Distribution Function (CDF)

Since the PDF is the derivative of the CDF:

$$
f(x)=F'(x)
$$

the CDF can be obtained from the PDF using integration:

$$
F(x)
=
\int_{-\infty}^{x}
f(u)\,du
$$



### Probability over an Interval

The probability that a continuous random variable $X$ lies within the interval:

$$
[a,b]
$$

is:

$$
P(a\leq X\leq b)
=
\int_{a}^{b}
f(x)\,dx
$$



This probability can be interpreted as:

- The area under the density curve
- The area above the horizontal axis
- The area between the limits $a$ and $b$

### What is a Characteristic Function (CF)?

In probability theory, the **Characteristic Function (CF)** of a random variable $X$ is a complete description of its probability distribution. Formally, for any random variable $X$, the characteristic function $\phi_X(t)$ is defined as the expected value of $e^{itX}$, where $i$ is the imaginary unit ($i^2 = -1$):

$$
\phi_X(t) = E\left[e^{itX}\right]
$$

For a continuous random variable with a probability density function $f(x)$, this is equivalent to the **Fourier transform** of the density function:

$$
\phi_X(t) = \int_{-\infty}^{\infty} e^{itx} f(x) \, dx
$$

For a discrete random variable taking values $x_k$ with probabilities $p_k$, it is defined as:

$$
\phi_X(t) = \sum_{k} p_k e^{itx_k}
$$



### Why does a CF always exist?

Unlike the **Moment Generating Function (MGF)**,

$$
M_X(t) = E[e^{tX}]
$$

which may not exist for many distributions (e.g., the Cauchy distribution) because the integral can diverge as $t$ grows, the Characteristic Function is **guaranteed to exist for every probability distribution.**

Here is why:

#### 1. The Bound on the Integrand

The definition of the CF involves the term $e^{itx}$. Using Euler's formula, we know that:

$$
e^{itx} = \cos(tx) + i \sin(tx)
$$

Because both the cosine and sine functions are bounded between $-1$ and $1$, the absolute value of the integrand is always bounded:

$$
\left|e^{itx}\right|
=
\left|\cos(tx) + i \sin(tx)\right|
=
\sqrt{\cos^2(tx) + \sin^2(tx)}
=
1
$$

#### 2. Convergence of the Expectation

Since the total probability mass of any distribution must sum (or integrate) to $1$, and the function being expected is bounded by $1$, the integral (or sum) is guaranteed to converge:

$$
\left|\phi_X(t)\right|
=
\left|E[e^{itX}]\right|
\leq
E\left[\left|e^{itX}\right|\right]
=
E[1]
=
1
$$

Because the absolute value of the characteristic function is always less than or equal to $1$ for all real $t$, it is well-defined and finite for every random variable, regardless of whether it has moments (such as the mean or variance) or how heavy its tails are.


### Key Properties

- **Uniqueness:** There is a one-to-one correspondence between a distribution and its CF. If two random variables have the same CF, they have the same distribution.
- **Independence:** The CF of a sum of independent random variables is simply the product of their individual CFs:

$$
\phi_{X+Y}(t) = \phi_X(t)\cdot\phi_Y(t)
$$

- **Moment Generation:** If the $n^{\text{th}}$ moment $E[X^n]$ exists, the CF is $n$-times differentiable at $t=0$, and the moment can be found via:

$$
E[X^n] = i^{-n}\phi_X^{(n)}(0)
$$

## 10.d.1. Introduction

The Negative Exponential Distribution (also known as the **Exponential Distribution**) is the probability distribution that describes the time between events in a Poisson process. *(Wikipedia – Exponential Distribution)*

A **Poisson process** is a type of random mathematical object known as a point process or point field that consists of randomly located points on some underlying mathematical space. *(Wikipedia – Poisson Point Process)*

- How much time will elapse before rainfall occurs in a given region?
- How long will it take before a call center receives the next phone call?
- How long will a vending machine work before it breaks down?

### Properties

- Exponential distribution refers to a statistical distribution used to model the time between independent events that happen at a constant average rate $\lambda$.
- $X$ (random variable), which corresponds to the distance between the successive counts of a Poisson process with mean $\lambda > 0$, is an exponential random variable with parameter $\lambda$.
- In the scenario, on average, the number of ice-creams sold per hour is $10$, i.e., $1$ customer per $6$ minutes. Therefore,

$$

$$

### Memory Less Property

The Negative Exponential Distribution has an important property called the **memory-less property**, and it can be defined as:

> "The future states of the process are independent of its past history and depend solely on its present state."

$$
P(X \ge t+h \mid X>t), \qquad t\ge0
$$

Here, $P(X \ge t+h \mid X>t)$ denotes the conditional probability of $X$ being larger than $t+h$, given that it is larger than $t$.

This relation implies that the probability that the event happens during a time interval of length $h$ is independent of how much time has already elapsed ($t$) without the event happening.

According to the memory-less property, even if $5$ minutes have already elapsed in an hour, the probability of customers not arriving in the next $7$ minutes is still **31.14%**.

A **joint probability distribution** is the probability distribution of two random variables, $X$ and $Y$, that defines their simultaneous behavior.

In the Joint Probability Distribution, $f(x,y)$, we use:

- **Joint PMF** to get the probability if $X$ and $Y$ are two discrete random variables.

# 15. Normality Tests

## 15.1. Introduction

A statistical test provides a mechanism for making quantitative decisions about a process or processes. The aim is to determine whether there is sufficient evidence to reject a null hypothesis.

A **normality test** is used to determine whether sample data is derived from a normally distributed population. More precisely, these tests are a form of model selection.

In statistics, a **goodness-of-fit test** is used to verify how well the observed data matches the expected model.

The suitability of a statistical model describes how well it fits a set of observations and summarizes the discrepancy between the observed values and the expected values under the model. The statistical model incorporates a set of assumptions regarding the generation of the observed data and similar data from a larger population.

### Types of Normality Tests

You will learn the following normality tests.

#### Graphical Tests

- Histogram
- Quantile-Quantile (Q-Q) Plot

#### Numerical Tests

- Chi-Square Test
- Shapiro-Wilk Test
- Kolmogorov-Smirnov Test

### Hypothesis Notations

In statistics, an assumption (or hypothesis) is tested based on observing a process that is modeled using a set of random variables.

This assumption is denoted by the **Null Hypothesis ($H_0$)**, and the **Alternative Hypothesis ($H_1$ or $H_a$)**.

The **Null Hypothesis** usually refers to a general statement or default position that there is no relationship between two measured phenomena. Rejecting the null hypothesis concludes that there is sufficient evidence to believe that there is a relationship between the two phenomena.

### 15.2.2. Quantile-Quantile (Q-Q) Plot

The **Quantile-Quantile (Q-Q) Plot** is a graphical tool used to assess whether a set of data plausibly comes from a theoretical distribution such as the **Normal Distribution** or **Exponential Distribution**.

A **45-degree reference line** is plotted to determine whether the two datasets come from populations with the same distribution. The greater the departure from this reference line, the stronger the evidence that the two datasets come from populations with different distributions.



image.png





#### What are Quantiles?

A **percentile** (also called a **quantile** or **centile**) is a measure used in statistics indicating the value below which a given percentage of observations in a dataset falls.

These are points in the data below which a certain proportion of the observations fall. Imagine a normally distributed bell curve with a mean of `0`. The `0.5` quantile (or `50th` percentile) is `0`, meaning half of the data lies below `0`.

Q-Q plots take your sample data, sort it in ascending order, and then plot it against quantiles calculated from a theoretical distribution. The number of quantiles is selected to match the size of the sample data.
