# 2. Measures of Central Tendency



## 2.1. Central Tendency

Central tendency measures are statistical measures used to describe or understand a sample using a single value.

This value could be the most typical value found in the dataset or the value that lies at the center of the data.

The measure of central tendency denotes the expected value in a given dataset and need not always be a value present in the dataset.

A measure of central tendency could also indicate the tendency of data to center around a particular value.

Commonly used measures of central tendency include:

- Arithmetic Mean
- Geometric Mean
- Harmonic Mean
- Weighted Mean
- Median
- Mode

### Uses of Central Tendency Measures

In the data collected from the survey, we can leverage the knowledge of measures of central tendency to answer questions such as:

- What would be the educational qualification of the people in this city?
- How much water does a household use in a month?
- How many patients does a doctor see in a particular working day?
- Which is the most common disease prevalent in this city during a particular time frame?
- What is the total monthly income of a middle-class or upper middle-class family in this city?



### Summary

- Measures of central tendency represent a single value that describes a sample.
- A measure of central tendency is the expected value in a dataset that represents the values in the entire dataset.
- Arithmetic Mean, Geometric Mean, and Harmonic Mean are used to determine the average or typical value of the data.
- The Weighted Mean is used when all data members do not contribute equally to the dataset.
- The Median represents the center of the data when the data is ordered and is computed only for ordered data. Unlike the average, it remains unaffected by the presence of extreme values (outliers) in the dataset.
- The Mode is a measure that describes the most frequently occurring value in a dataset. It is commonly used for non-numeric data (though it is not limited to non-numeric data).

You can use the following commands to compute the values of **mean**, **median**, and **mode** as shown below.

````python
import statistics as st
print(st.mean([1,2,3,4]))
print(st.median([7800,8000,7500,7300,7000,6500,6750,90000,8000,7800]))
print(st.mode([1,2,2,3,4,4,4]))

````

## 2.2. Arithmetic Mean

The **Arithmetic Mean** (generally known as the **average**) is a widely used measure to describe a population using a single value.

Arithmetic Mean is a measure used for **numeric data** and is a single value that represents the typical value in a given set of numbers.

**Example:**

- Mean marks of a class = 70 out of 100.

In the examples described earlier, the Arithmetic Mean is used to answer questions such as:

- How much water does a household in this city use in a month?
- How many patients does a doctor in the city see in a day?

### Computing Arithmetic Mean

The Arithmetic Mean is the sum of all values in the dataset divided by the total number of values in the dataset.




image.png




Depending on whether the data being analyzed is a **sample** or the **population**, the notation used to represent the mean changes.

- Sample Mean: $\bar{x}$
- Population Mean: $\mu$

The Arithmetic Mean can be computed using the `mean()` function from the `statistics` module in Python, as shown below.

````python
# Importing the statistics module
import statistics
# list of positive integer numbers
data1 = [1, 3, 4, 5, 7, 9, 2]
x = statistics.mean(data1)
# Printing the mean
print("Mean is :", x)
````

## 2.3. Weighted Mean

The **Weighted Mean** is used when some values in the dataset contribute more than others.

We can calculate the Weighted Mean by dividing the sum of the products of each value and its corresponding weight by the total weight.



image.png



### Need

During the survey, **100 samples** of water were collected and sent to two different laboratories for analysis.

- **Lab 1** analyzed **30 samples** and reported **40 ppm** of harmful substances.
- **Lab 2** analyzed **70 samples** and reported **63 ppm** of harmful substances.

The average computed using the Arithmetic Mean would be:

$$
\frac{40+63}{2}=51.5\ \text{ppm}
$$

This value, however, does not truly represent the average because it does not consider the difference in the sizes of the datasets.

To account for the unequal contribution of each laboratory, we use the **Weighted Mean**.

Using the Weighted Mean results in an average of **56.1 ppm** of harmful substances.

The Weighted Mean can be computed using `np.average()` in Python, as shown below.


````python
#weighted average
import numpy as np
print("arithmetic mean:", np.mean([1,2,3,4]))#arithmetic mean
print("arithmetic mean:", np.average([1,2,3,4]))#arithmetic mean
print("weighted mean:", np.average([40,63], weights=[30,70]))# weighted mean
````

# 3. Measures of Dispersion

## 3.1. Dispersion

Dispersion represents the **variability** in the data. In simple terms, **Measures of Dispersion** indicate how squeezed or stretched the data is in comparison with the measures of central tendency.

In other words, Measures of Dispersion help us identify how the data is spread across the dataset.

A measure of dispersion is a **non-negative real number** that increases as the diversity (spread) of the data increases.

Commonly used measures of dispersion include:

- Range
- Variance
- Standard Deviation
- Interquartile Range (IQR)

### Example

Consider two different communities in the city, each having **5 doctors**.

So, what is the alternative?

The answer is **Measures of Dispersion**.

## 3.2. Range

The **Range** is the simplest measure of dispersion and provides an idea of how much the data varies within a dataset.

$$

### Chebyshev's Inequality

Chebyshev's Inequality relates the **Mean** and the **Standard Deviation**. It provides an upper bound on the probability of finding a value that is far from the mean.

This implies:

- Values closer to the mean have a **higher probability** of occurring.
- As a value moves farther away from the mean, the probability of observing it decreases.

The inequality is expressed as:

$$
P\left(|X-\mu|\geq k\sigma\right)\leq\frac{1}{k^2}
$$

This means:

> If a value is **$k$ standard deviations away from the mean**, then the probability of finding that value is **less than or equal to $\frac{1}{k^2}$**.

## 3.5. Z-Score

Comparison of data from different datasets is performed by observing how a value compares with the mean of the dataset from which it originates.

This comparison can be performed using the **Z-score**, which indicates the distance of a data value from the mean in terms of **Standard Deviations**.

The Z-score of a value $x$ is computed using the formula:

$$
z=\frac{x-\mu}{\sigma}
$$

where:

- $x$ = Data value
- $\mu$ = Mean
- $\sigma$ = Standard Deviation

## Example

Doctor 1 graduated from **University A** in the year **2015** with an aggregate of **78%**, whereas Doctor 2 graduated from **University B** in the year **2015** with an aggregate of **76%**.

Both doctors apply for a position at a government hospital in your city. The recruiter needs to compare their academic performance to decide which candidate to select.

The following information is available:

| University | Mean ($\mu$) | Standard Deviation ($\sigma$) |
|------------|-------------:|------------------------------:|
| University A | 72 | 4 |
| University B | 68 | 3 |

How can we compare data from these different datasets?

A Z-score of:

- **2** indicates that the value is **2 Standard Deviations above the mean**.
- **-1.3** indicates that the value is **1.3 Standard Deviations below the mean**.

### Calculation

For **Doctor 1**:

$$
z=\frac{78-72}{4}=1.5
$$

For **Doctor 2**:

$$
z=\frac{76-68}{3}=2.33
$$

Since **Doctor 2** has the higher Z-score, Doctor 2 performed better relative to other students in their respective university and is therefore the more suitable candidate in this scenario.

## 3.6. Quartiles / Percentiles

The **$n$th Percentile** is identified using the following procedure:

1. Arrange all the data in **ascending order**.
2. Multiply the size of the dataset by $n$ (the value obtained is called the **index**).
3. Determine the percentile value using the index:
   - If the **index is not a whole number**, round it to the nearest whole number and identify the value at that index. This is the percentile value.
   - If the **index is a whole number**, the percentile value is the average of the values at the **index** and **index + 1**.

### Quartiles

Quartiles are **three points** in a dataset that divide it into **four equal groups**.

- **First Quartile ($Q_1$):** The **25th percentile**, i.e., the value below which **25%** of the data lies.
- **Second Quartile ($Q_2$):** The **50th percentile**, i.e., the value below which **50%** of the data lies. This is also the **Median ($M$)** of the dataset.
- **Third Quartile ($Q_3$):** The **75th percentile**, i.e., the value below which **75%** of the data lies.

### Example

A health officer wants to determine how busy the doctors in a community are. Specifically, she wants to identify how many patients the **busiest 20% of doctors** attend to in a day.

The following data represents the average number of patients attended per day by **20 doctors** in a community.

A **Percentile** represents the value below which a given percentage of observations falls.

In this example, we need to calculate the **80th percentile**, which is:

- The value below which **80%** of the observations lie, or
- The value above which **20%** of the observations lie.

Using this procedure, the **80th percentile** of the given dataset is **11.5**.

It is expressed as:

$$

$$

The Interquartile Range is used to indicate the spread of the data and is generally a better measure than the **Range** because the minimum and maximum values (used in the Range) are highly sensitive to unusual or extreme values.

For the example discussed earlier:

$$
IQR = Q_3 - Q_1 = 11 - 6 = 5
$$

````python
#interquartile range
from scipy.stats import iqr
x = [8,12,4,11,15,10,8,7,4,11,6,6,8,7,4,9,16,6,14,6]
iqr(x)
````

## 4.2. Skewness

**Skewness** is the measure of the lack of symmetry of a dataset in comparison with a symmetric distribution.

In a **symmetrically distributed** dataset, the **Mean** and the **Median** coincide.

As discussed earlier, the presence of unusually large or small values (called **outliers**) affects the Mean. The shift of the Mean away from the Median depends on the nature of these unusual values.

- If there are **unusually large values**, the **Mean becomes greater than the Median**, indicating a **right-skewed** distribution.
- If there are **unusually small values**, the **Mean becomes less than the Median**, indicating a **left-skewed** distribution.

The Skewness of a dataset is calculated using the following formula:

$$
\text{Skewness}=\frac{3(\text{Mean}-\text{Median})}{\text{Standard Deviation}}
$$

Skewness helps us determine whether the **Mean** is an appropriate measure of central tendency for a dataset.

In the case of **Community 1** and **Community 2**, the Mean and Median are very close, indicating that the **Mean** is an acceptable measure of central tendency.

The Skewness can be computed in Python using the following code:





````python
#skewness
from scipy.stats import skew
skew([11730,5461,6655,7484,10242,8547,5521,13207,7598,9421,11563,6995,10531,11525,7770,7094,8339,8375,8808,6802])
````

# 6. Sampling

## 6.1. Introduction

### Challenge 1: How many students should be selected for the surprise test?

Suppose the surveying agency decides to conduct the surprise test on a selected group of students.

The agency must determine the number of students who should appear for the surprise test so that they adequately represent the performance of all students in the country.

In statistical terms, the agency must determine the **Sample Size**.

The **Sample Size** is determined by the desired **Level of Accuracy**.

The Level of Accuracy indicates how closely the estimated population parameters are expected to match the actual population parameters.

The Level of Accuracy is measured in terms of:

- Confidence Level
- Maximum Permissible Error

The following sections describe how these parameters affect the Sample Size.

## 6.2. Sample Size

### 6.2.1. Factors Affecting Sample Size: Confidence Level

The surveying agency estimates the quality of primary school education by analyzing the performance of a selected sample of students.

Suppose the agency estimates that the average performance of primary school students is **72%**.

Since this is a **point estimate**, it may not exactly match the true population value.

Instead, the agency may estimate that the average performance lies between **70% and 75%**.

This interval is called the **Confidence Interval**, and the certainty associated with this interval is measured by the **Confidence Level**.

The **Confidence Level** represents the probability that the population parameter lies within the Confidence Interval.

For example:

- **95% Confidence Level** implies that the probability of the population parameter lying within the confidence interval is **0.95**.

To achieve a higher level of confidence, more observations are generally required.

Therefore:

- **Higher Confidence Level ⇒ Larger Sample Size**



### 6.2.2. Factors Affecting Sample Size: Maximum Permissible Error

Estimated parameters differ from the true population parameters because only a sample, rather than the entire population, is analyzed.

Accurate analysis requires that this difference be as small as possible.

The **Maximum Permissible Error** specifies the largest error that can be tolerated in the analysis.

To reduce the Maximum Permissible Error, a larger sample is generally required.

The relationship between Sample Size and Maximum Permissible Error is illustrated below:



image.png




Therefore:

- Increasing the **Confidence Level** increases the required Sample Size.
- Decreasing the **Maximum Permissible Error** also increases the required Sample Size.



### 6.2.3. Factors Affecting Sample Size: Population Variance

Another important factor affecting Sample Size is the **Population Variance**.

Consider two different populations shown below.




image.png





From the observations:

- **Population 1** has **low variance**, meaning the observations are close to one another.
- **Population 2** has **high variance**, meaning the observations are widely spread.

Now, suppose three samples of size **2** are selected from both populations, and their sample means are calculated.

#### Population 1

The sample means are:

- 11

These values are close to the **Population Mean (11)**.

Since the population variance is low, even a small sample size adequately represents the population.

#### Population 2

The sample means are:

- 5
- 15
- 17.5

These values differ significantly from the **Population Mean (10)**.

Since the population variance is high, a small sample may fail to represent the population because it may not include the extreme values responsible for the high variance.

Therefore, populations with higher variance generally require larger sample sizes.

Now, increase the sample size to **4** for Population 2.




image.png





The new sample means are much closer to the Population Mean.

Hence, these samples are considered better representatives of the population.

Therefore:

- **Higher Population Variance (or Standard Deviation) ⇒ Larger Sample Size**


### 6.2.4. Determining Sample Size

The Sample Size depends primarily on the following three factors:

- Confidence Level
- Maximum Permissible Error
- Population Variance (or Standard Deviation)

The Sample Size can be calculated using the following formula:

| $z_{\text{critical}}$ | Critical Z-statistic for the specified Confidence Level |
| $\sigma$ | Population Standard Deviation |
| $E$ | Maximum Permissible Error |

## 7.2. Simple Random Sampling

Assume that, from a population of size $N$, all possible samples of size $n$ are generated.

If every sample has an equal probability of being selected, the process is called **Simple Random Sampling**.

From a random sample, we understand that:

- A random sample of size $n$ is a set of $n$ observations selected from a **finite** or **infinite** population.

### Examples

| Population Type | Example |
|-----------------|---------|
| Finite Population | Number of students in a country |
| Infinite Population | Number of times a coin can be tossed |

Assume:

- The population size is $N$.
- Every subset of size $n$ has an equal probability of being selected.

A random sample of size $n$, denoted by $(X_1, X_2, \ldots, X_n)$, selected from an infinite population having probability density function $f(x)$, satisfies the following:

- Each $X_i$ is a random variable with distribution $f(x)$.
- The random variables are mutually independent.

### Simple Random Sampling in Python

One of the most common methods of Simple Random Sampling is the **Lottery Method**.

Each of the $N$ observations is assigned a unique identifier (e.g., `s0`, `s1`, ..., `s15`).




### Drawbacks of Simple Random Sampling

In Simple Random Sampling, the selection is based entirely on chance.

As a result, the selected sample may not accurately represent the characteristics of the entire population.

For example, the **1120 students** selected by the surveying agency may coincidentally all be high performers.

Such a sample would not accurately represent the overall quality of primary education and may produce **biased** conclusions.

A **Biased Sample** is a sample that does not properly represent the population.

**Bias** refers to the tendency of a statistic to systematically **overestimate** or **underestimate** a population parameter.

To avoid such issues, the surveying agency chooses **Stratified Sampling**.

### Challenge 3.1: Estimating Population Mean

### 8.5.1. Point Estimate of Population Mean

The surveying agency begins by estimating the **Population Mean**.

Estimating the Population Mean involves:

- Finding the **Point Estimate** of the Population Mean, i.e., estimating the mean marks that would have been obtained if **all students** in the country had appeared for the surprise test.
- Finding the **95% Confidence Interval** for the Population Mean, i.e., determining an interval within which the true Population Mean lies with **95% confidence**.

Let us first understand how the Point Estimate of the Population Mean is obtained.

The Point Estimate of the Population Mean is determined in two steps.

### Step 1: Finding an Estimator

The surveying agency identifies a **Sample Statistic** that closely represents the Population Mean.

- The Sample Statistic is called the **Estimator** of the Population Parameter.
- The numerical value obtained from the estimator is called the **Estimate** of the Population Mean.

### Step 2: Evaluating the Estimator

After selecting an estimator, the agency evaluates its quality.

One of the most commonly used measures of quality is the **Unbiasedness** of the estimator.

Suppose the expected value of the estimator is calculated.

The **Bias** of the estimator is defined as:

$$
\text{Bias} = E(\hat{\theta}) - \theta
$$

where:

- $\hat{\theta}$ = Estimator
- $\theta$ = True Population Parameter

If the Bias is **zero**, the estimator is said to be **Unbiased**.

Once the estimator is found and verified to be unbiased, it is considered an **Unbiased Estimator** of the Population Parameter.

Several methods exist for identifying suitable estimators.

One of the most widely used methods is **Maximum Likelihood Estimation (MLE)**.

Therefore, the surveying agency uses the **MLE** method.

### 8.5.2. Maximum Likelihood Estimation (MLE)

**Maximum Likelihood Estimation (MLE)** is a statistical method used to estimate unknown population parameters from observed sample data.

It involves:

1. Constructing the **Likelihood Function** based on the assumed probability distribution.
2. Determining the values of the unknown parameters that maximize the Likelihood Function.

Since the students' marks are assumed to follow a **Normal Distribution**, the agency derives the likelihood function for normally distributed observations.

#### Likelihood Function for a Normally Distributed Variable

The **Probability Density Function (PDF)** of a statistical model, denoted by:

$$
f(x \mid \theta)
$$

represents the probability density of observing the data vector $x$ given the population parameter $\theta$.

The **Likelihood Function**, denoted by:

$$
L(\theta \mid x)
$$

represents the likelihood of observing the population parameter $\theta$ given the observed sample $x$.

In other words:

- **PDF:** Probability of observing the data when the population parameter is known.
- **Likelihood Function:** Likelihood of the unknown population parameter given the observed data.
